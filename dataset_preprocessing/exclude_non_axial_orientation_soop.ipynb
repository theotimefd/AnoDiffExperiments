### Search for images with bad spacing to exclude them
e.g images that were acquired with thick coronal or sagittal slices instead of thick axial slices

```python
bad_images_names = set()
count = 0

for fp in soop_flair_images:

    try:
        img = nib.load(fp)
        zooms = img.header.get_zooms()
        spacing = zooms[:3] if len(zooms) >= 3 else zooms
        x, y, z = spacing
        x = np.round(x, 2)
        y = np.round(y, 2)
        z = np.round(z, 2)
        if x>1.0 or y>1.0:
            #print("bad image")
            bad_images_names.add(os.path.basename(fp).split('_')[0])
            """ count +=1
            print(f"{os.path.basename(fp)}: spacing (mm) = {[x, y, z]}")
            
            data = img.get_fdata()
            print(data.shape)

            if data.ndim == 4:
                data = data[..., 0]
            k = data.shape[2] // 2
            slice_axial = data[:, :, k]

            plt.figure(figsize=(5, 5))
            plt.imshow(slice_axial, cmap='gray', origin='lower')
            plt.title(f"{os.path.basename(fp)} | axial slice z={k}")
            plt.axis('off')
            plt.show() """
    except Exception as e:
        print(f"Failed to read {fp}: {e}")
```

```python
# Write bad_images_names to a CSV file (one entry per line)
out_csv = os.path.join(ROOT_DIR, "AnoDiffExperiments/data_splits_lists/final_soop_dataset_small/exclude_non_axial_thick_slices.csv")
names_sorted = sorted(bad_images_names)
np.savetxt(out_csv, np.array(names_sorted, dtype=str), fmt='%s', newline='\n')
print(f"Saved {len(names_sorted)} entries to {out_csv}")
```
