### AINI-Stroke AIT for 2022-2023
AIT (Accident ischemique transitoire) = Not visible on ADC ("healthy ADC")

```python
clinical_data_2022 = pd.read_csv("/bettik/PROJECTS/pr-gin5_aini/fehrdelt/datasets/donnees_cliniques/Clinical_data_2022.csv")
clinical_data_2023 = pd.read_csv("/bettik/PROJECTS/pr-gin5_aini/fehrdelt/datasets/donnees_cliniques/Clinical_data_2023.csv")
```

```python
correspondance_noms_2022 = pd.read_csv("/bettik/PROJECTS/pr-gin5_aini/fehrdelt/datasets/donnees_cliniques/correspondance_noms_2022.csv", names=["n°patient", "Shanoir_name"])
correspondance_noms_2023 = pd.read_csv("/bettik/PROJECTS/pr-gin5_aini/fehrdelt/datasets/donnees_cliniques/correspondance_noms_2023.csv", names=["n°patient", "Shanoir_name"])
```

```python
correspondance_noms_2022.head()
```

```python
clinical_data_2022.head()
```

```python
# list all columns
clinical_data_2022["type_imagerie"]
```

```python
clinical_data_2022 = pd.merge(clinical_data_2022, correspondance_noms_2022, on="n°patient", how='left')
clinical_data_2023 = pd.merge(clinical_data_2023, correspondance_noms_2023, on="n°patient", how='left')
```

```python
AITs_2022 = clinical_data_2022.loc[clinical_data_2022["diagnostic"]=="AIT"]["Shanoir_name"]
AITs_2022_and_IRM = clinical_data_2022.loc[(clinical_data_2022["diagnostic"]=="AIT") & (clinical_data_2022["type_imagerie"]=="IRM")]["Shanoir_name"]

AITs_2023 = clinical_data_2023.loc[clinical_data_2023["diagnostic"]=="AIT"]["Shanoir_name"]
AITs_2023_and_IRM = clinical_data_2023.loc[(clinical_data_2023["diagnostic"]=="AIT") & (clinical_data_2023["type_imagerie"]=="IRM")]["Shanoir_name"]
```

```python
names_AITs = AITs_2022_and_IRM.to_list() + AITs_2023_and_IRM.to_list()
print(len(names_AITs))
```

### AINI-Stroke AIT for 2018-2021
AIT (Accident ischemique transitoire) = Not visible on ADC ("healthy ADC")

```python
clinical_data_2018_2023 = pd.read_csv("/bettik/PROJECTS/pr-gin5_aini/fehrdelt/datasets/donnees_cliniques/AINI-Stroke_2018-2021_filtered_05-2023.csv")
```

```python
clinical_data_2018_2023.head()
```

```python
AITs_2018_2021 = clinical_data_2018_2023.loc[clinical_data_2018_2023["@32_diag_2"]==2]["id_data"]

AITs_2018_2021_and_IRM = clinical_data_2018_2023.loc[(clinical_data_2018_2023["@32_diag_2"]==2) & (clinical_data_2018_2023["@38_imagerie"]==1)]["id_data"]
```

```python
names_AITs = AITs_2018_2021_and_IRM.to_list()
print(len(names_AITs))
```

```python
folder_to_process = "/bettik/PROJECTS/pr-gin5_aini/fehrdelt/datasets/Aini-Stroke_ADC/Others_2018_2021/"
AIT_folder_2018_2021 = "/bettik/PROJECTS/pr-gin5_aini/fehrdelt/datasets/Aini-Stroke_ADC/AIT_2018_2021/"

for file in os.listdir(folder_to_process):
    if file.split('_')[0] in names_AITs:
        os.rename(os.path.join(folder_to_process, file), os.path.join(AIT_folder_2018_2021, file))
```
