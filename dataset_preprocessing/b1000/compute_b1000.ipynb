```python
import os
import glob

import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm

import nibabel as nib
```

```python
ROOT_DIR = "/home/fehrdelt/bettik/"
#ROOT_DIR = "/bettik/PROJECTS/pr-gin5_aini/fehrdelt/"
```

```python
ixi_folder = ROOT_DIR+ "datasets/IXI-DTI/"
dallas_folder = ROOT_DIR+ "datasets/dwi_dallas/"
```

```python
nii_files = glob.glob(os.path.join(dallas_folder, "*.nii.gz"))
```

```python
print(len(nii_files))
```

```python
img = nib.load(nii_files[0])
data = img.get_fdata()
print(data.shape)
```

```python
for file in tqdm(nii_files):
    
    image_nifti = nib.load(file)
    image_data = image_nifti.get_fdata()
    
    b1000_map = np.mean(image_data[...,1:], axis=-1)

    b1000_nifti = nib.Nifti1Image(b1000_map, image_nifti.affine, image_nifti.header)
    save_path = os.path.join(f'{ROOT_DIR}/datasets/b1000_dallas/{os.path.basename(file)}')
    nib.save(b1000_nifti, save_path)

    #print(f"Saved B1000 map for patient {patient_id} to {save_path}")
```
