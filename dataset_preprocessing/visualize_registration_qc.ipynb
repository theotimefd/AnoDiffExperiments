```python
final_t1_dataset_small_ixi = ROOT_DIR+"datasets/final_t1_dataset_small/ixi_registered/"
final_t1_dataset_small_oasis = ROOT_DIR+"datasets/final_t1_dataset_small/oasis_registered/"
final_t1_dataset_small_dallas = ROOT_DIR+"datasets/final_t1_dataset_small/dallas_registered/"
```

```python
template_t1_path = ROOT_DIR + "datasets/mni_icbm152_nlin_sym_09a/template_t1_128.nii.gz"
```

```python
oasis_t1_qc = registration_qc.get_nifti_list_by_registration_quality(final_t1_dataset_small_oasis, template_t1_path, num_workers=32)
```

```python
start_index = 500
num_images = 64
grid_size = 8

fig, axes = plt.subplots(grid_size, grid_size, figsize=(20, 20))
axes = axes.flatten()

for i in range(num_images):
    curr_index = start_index + i
    filename = oasis_t1_qc[curr_index]
    file_path = os.path.join(final_t1_dataset_small_oasis, filename)
    
    try:
        img = nib.load(file_path)
        data = img.get_fdata()
        
        # Get the middle axial slice
        middle_slice = data[:, :, data.shape[2] // 2]
        
        axes[i].imshow(middle_slice.T, cmap='gray', origin='lower')
        
        axes[i].set_title(f"Idx: {curr_index}", fontsize=10)
    except Exception as e:
        axes[i].set_title("Error loading", fontsize=10)
        
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
